```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Tell Python where your data folder is
import os

data_folder = os.path.expanduser("~/Desktop/gcc-wef-nexus/data")

print("Looking for files in:", data_folder)
print("\nFiles found:")
for file in os.listdir(data_folder):
    print(" -", file)
```

```python
# Load the World Bank data
wb = pd.read_csv(os.path.join(data_folder, "worldbank_gcc.csv"), skiprows=4)

# Show the first 5 rows
print("Shape:", wb.shape)
print("\nFirst 5 rows:")
wb.head()
```

```python
# See all column names clearly
print("All column names:")
for col in wb.columns:
    print(" -", col)
```

```python
# See all 6 countries and their indicators
print("Countries in your data:")
print(wb.iloc[:, 0].unique())

print("\nIndicators in your data:")
print(wb.iloc[:, 2].unique())
```

```python
# Rename the first 4 columns to simple names
wb.columns = ['country', 'code', 'indicator', 'indicator_code'] + list(wb.columns[4:])

# Confirm the rename worked
print("New column names (first 8):")
print(list(wb.columns[:8]))

print("\nData shape:", wb.shape)
print("\nCountries:")
print(wb['country'].unique())
print("\nIndicators:")
print(wb['indicator'].unique())
```

```python
# Remove the indicator code column - we don't need it
wb_clean = wb.drop(columns=['indicator_code'])

# Melt the year columns into rows
# This transforms from wide format to long format
wb_long = wb_clean.melt(
    id_vars=['country', 'code', 'indicator'],
    var_name='year',
    value_name='value'
)

# Convert year to a number
wb_long['year'] = pd.to_numeric(wb_long['year'], errors='coerce')

# Convert value to a number
wb_long['value'] = pd.to_numeric(wb_long['value'], errors='coerce')

# Remove empty rows
wb_long = wb_long.dropna(subset=['year'])

print("Clean data shape:", wb_long.shape)
print("\nFirst 10 rows of your clean data:")
wb_long.head(10)
```

```python
# Create a short name for each indicator - easier to work with
indicator_names = {
    'Food imports (% of merchandise imports)': 'food_imports',
    'Prevalence of undernourishment (% of population)': 'undernourishment',
    'Access to electricity (% of population)': 'electricity_access',
    'Renewable internal freshwater resources per capita (cubic meters)': 'freshwater_per_capita',
    'Annual freshwater withdrawals, total (% of internal resources)': 'water_withdrawal',
    'Renewable electricity output (% of total electricity output)': 'renewable_electricity'
}

# Apply the short names
wb_long['indicator'] = wb_long['indicator'].map(indicator_names)

# Check all indicators mapped correctly
print("Indicator names after mapping:")
print(wb_long['indicator'].unique())
```

```python
# Pivot: each indicator becomes its own column
wb_pivot = wb_long.pivot_table(
    index=['country', 'code', 'year'],
    columns='indicator',
    values='value'
).reset_index()

# Clean up column naming
wb_pivot.columns.name = None

print("Master table shape:", wb_pivot.shape)
print("\nColumn names:")
print(list(wb_pivot.columns))
print("\nFirst 5 rows:")
wb_pivot.head()
```

```python
# Read the raw file with NO skipping to see its true structure
wb_raw = pd.read_csv(
    os.path.join(data_folder, "worldbank_gcc.csv"), 
    header=None, 
    nrows=8
)

# Show every row and column
pd.set_option('display.max_columns', 10)
pd.set_option('display.width', 200)
print(wb_raw)
```

```python
# Reload correctly - row 0 IS the header, no skipping needed
wb = pd.read_csv(os.path.join(data_folder, "worldbank_gcc.csv"))

# Rename first 4 columns
wb.columns = ['country', 'code', 'indicator', 'indicator_code'] + list(wb.columns[4:])

# Check year column names - they look like "2000 [YR2000]"
print("Year column examples:")
print(list(wb.columns[4:8]))
```

```python
# Melt into long format
wb_long = wb.drop(columns=['indicator_code']).melt(
    id_vars=['country', 'code', 'indicator'],
    var_name='year',
    value_name='value'
)

# Extract JUST the 4-digit year from "2000 [YR2000]"
wb_long['year'] = wb_long['year'].str.extract(r'(\d{4})').astype(float)

# Convert value to number
wb_long['value'] = pd.to_numeric(wb_long['value'], errors='coerce')

# Drop empty rows
wb_long = wb_long.dropna(subset=['year'])

print("Shape:", wb_long.shape)
print("\nYear values (first 5):")
print(wb_long['year'].head())
print("\nSample data:")
wb_long.head(10)
```

```python
# Apply short indicator names
indicator_names = {
    'Food imports (% of merchandise imports)': 'food_imports',
    'Prevalence of undernourishment (% of population)': 'undernourishment',
    'Access to electricity (% of population)': 'electricity_access',
    'Renewable internal freshwater resources per capita (cubic meters)': 'freshwater_per_capita',
    'Annual freshwater withdrawals, total (% of internal resources)': 'water_withdrawal',
    'Renewable electricity output (% of total electricity output)': 'renewable_electricity'
}

wb_long['indicator'] = wb_long['indicator'].map(indicator_names)

# Check mapping worked
print("Indicators after mapping:")
print(wb_long['indicator'].unique())
```

```python
# Rebuild the pivot table
wb_pivot = wb_long.pivot_table(
    index=['country', 'code', 'year'],
    columns='indicator',
    values='value'
).reset_index()

wb_pivot.columns.name = None

# Convert year to integer
wb_pivot['year'] = wb_pivot['year'].astype(int)

print("Master table shape:", wb_pivot.shape)
print("\nColumns:", list(wb_pivot.columns))
print("\nYears range:", wb_pivot['year'].min(), "to", wb_pivot['year'].max())
print("\nCountries:", wb_pivot['country'].unique())
print("\nFirst 5 rows:")
wb_pivot.head()
```

```python
# Save your clean master table permanently
wb_pivot.to_csv(
    os.path.join(data_folder, "gcc_master_clean.csv"), 
    index=False
)

print("✓ Master table saved successfully!")
print("✓ Location: gcc-wef-nexus/data/gcc_master_clean.csv")
print(f"✓ {wb_pivot.shape[0]} rows × {wb_pivot.shape[1]} columns")
print("\nPhase 2 is complete. Your data is clean and ready.")
```
